# MC-CAST: exploration

Run `mccast all` first (or the individual steps). This notebook inspects the results.

```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mccast import config as C
```

## Area by class over time

```python
area = pd.read_csv(C.OUTPUTS / 'area_by_class.csv')
area.pivot_table(index='year', columns='class', values='area_km2').plot(marker='o', figsize=(8, 4))
plt.ylabel('km²');
```

## Validation metrics (2015 → 2023)

```python
json.loads((C.OUTPUTS / 'metrics.json').read_text())
```

## Suitability lift per bin

```python
s = json.loads((C.OUTPUTS / 'suitability.json').read_text())
for name, f in s['factors'].items():
    plt.plot(f['lifts'], marker='o', label=name)
plt.axhline(1, color='k', lw=.5); plt.xlabel('quantile bin (near → far)'); plt.ylabel('lift'); plt.legend();
```
